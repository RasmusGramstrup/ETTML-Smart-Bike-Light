```python
import os
import csv
import pandas as pd

input_dir = './out/15_05_2/'

for file in os.listdir(input_dir):
    count = 1
    with open(input_dir + file, 'r', encoding="utf-8") as inp:
        lines = list(csv.reader(inp))
        print(lines[1])
        

        while True:
            fileName = ' '.join(map(str,lines[1])) + '.' + format(count,'05d') + '.csv'
            filePath = "./out_converted/15_05_test"
            path = os.path.join(filePath, fileName)
            print(path)
            if os.path.exists(path):
                count += 1
                continue
            else:
                break
        inp.close()

        df = pd.read_csv(input_dir + file)
        df =  df.drop(df.index[0])
        
        df1 = df.iloc[0::4, :]
        df2 = df.iloc[1::4, :]
        df3 = df.iloc[2::4, :]
        df4 = df.iloc[3::4, :]

        df1.to_csv(path, index=False)

        count += 1

        fileName = ' '.join(map(str,lines[1])) + '.' + format(count,'05d') + '.csv'
        path = os.path.join(filePath, fileName)

        df2.to_csv(path, index=False)

        count += 1

        fileName = ' '.join(map(str,lines[1])) + '.' + format(count,'05d') + '.csv'
        path = os.path.join(filePath, fileName)

        df3.to_csv(path, index=False)
        
        count += 1

        fileName = ' '.join(map(str,lines[1])) + '.' + format(count,'05d') + '.csv'
        path = os.path.join(filePath, fileName)

        df4.to_csv(path, index=False)
        
        # with open('./out/05_05/' + file, 'r', encoding="utf-8") as inp, open(path, 'w', encoding="utf-8") as output:
        #     writer = csv.writer(output)
        #     for row in csv.reader(inp):
        #         if row == lines[1]:
        #             continue
        #         else:
        #             writer.writerow(row)
                   

    # inp.close()
    # output.close()
        
    
    
```
